supervised classification method that seperates data using hyperplanes

Goal:
used for both classification as well as the regression problems
to create the best line or the decision boundary that can segregate n dimensional space into classes so that we can easily put the new data point in correct category

best decison boundary is hyperplanes

there can be multiple lines / decision boundaries that can seperates n dimensional space

in layman terms , we are separating classes by a line , a line can be drown in diffreent plane , but which one should be there ?

support vectors=> that data points which are on boundary and touching the hyperplane in both +ive and -ive sids,
distance between the support vectors or the margins should be maximum so that both classes can be clearly seperated and 
the new coming data can be easily putted into the required category

dimension of hyperplane depend on feature of dataset means if 2 features then hyperplane will be a staright line.
if 3 features, hyperplane will be 2 dimensional plane.

```python
df = pd.read_csv('wine.csv')
X = df.drop('Wine',axis=1)
y = df['Wine']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
linear_model = SVC(kernel = 'linear')
linear_model.fit(X_train,y_train)
linear_pred = linear_model.predict(X_test)
print('-------Linear SVM--------')
print('Accuracy:',accuracy_score(y_test,linear_pred))
print('\nClassification Report:\n',classification_report(y_test,linear_pred))
print('\nConfusion Matrix:\n',confusion_matrix(y_test,linear_pred))
```

```python
rbf_model = SVC(kernel = 'rbf')
rbf_model.fit(X_train,y_train)
rbf_pred = rbf_model.predict(X_test)
print('-------RBF SVM--------')
print('Accuracy:',accuracy_score(y_test,rbf_pred))
print('\nClassification Report:\n',classification_report(y_test,rbf_pred))
print('\nConfusion Matrix:\n',confusion_matrix(y_test,rbf_pred))
```
