```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, accuracy_score, classification_report, confusion_matrix
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv("spam.csv",encoding='latin1')
df
```

```python
df.groupby('Category').describe()
```

```python
df['spam'] = df['Category'].apply(lambda x:1 if x=='spam' else 0)
```

```python
df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.Message, df.spam, test_size=0.25)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
X_train_count = cv.fit_transform(X_train.values)
X_train_count.toarray()[:3]
```

```python

from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train_count,y_train)
```

```python
X_test_count = cv.transform(X_test.values)
y_pred = model.predict(X_test_count)

print("Accuracy score: ",accuracy_score(y_test,y_pred))
print("Confusion Matrix: \n",confusion_matrix(y_test,y_pred))
```

```python
emails = [
    'Hey mohan, can we get together to watch the cricket match tomorrow',
    "upto 20% discount on parking, exclusive offer is going on, don't miss that reward!"
]
email_count = cv.transform(emails)
model.predict(email_count)
```

```python
from sklearn.pipeline import Pipeline

clf = Pipeline([
    ('vectorizer',CountVectorizer()),
    ('nb',MultinomialNB())
])

clf.fit(X_train,y_train)
clf.score(X_test,y_test)
```

```python
from sklearn.datasets import load_wine
df = load_wine()

X = df.data
y = df.target
target_name = df.target_names
print(target_name)
print('\n')
print(X)
print('\n')
print(y)

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=0)
print('\n')
print(len(X_train))

print(len(X_test))

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(X_train,y_train)
y_pred = nb.predict(X_test)
acc_gnb = accuracy_score(y_test,y_pred)
print("Accuracy: ",acc_gnb)
print("Classification Report: \n",classification_report(y_test,y_pred))
print("Actual values: ",y_test[:5])
print("Predicted values: ",y_pred[:5])
print('\n Multinomial Naive Bayes workflow is: \n')
X_tr,X_ts,y_tr,y_ts = train_test_split(X,y,test_size=0.3,random_state=0)
ms = MinMaxScaler()

X_tr = ms.fit_transform(X_tr)
X_ts = ms.transform(X_ts)

model = MultinomialNB()
model.fit(X_tr,y_tr)
y_pre_ms = model.predict(X_ts)
acc_mnb = accuracy_score(y_ts,y_pre_ms)
print("Accuracy score: ",acc_mnb)
print("Classification report: \n",classification_report(y_ts,y_pre_ms))
print("Actual values: ",y_ts[:5])
print("Predicted values: ",y_pre_ms[:5])

if acc_gnb > acc_mnb:
    best_mdl = nb
    best_accuracy = acc_gnb
    model_name = "Gaussian Naive Bayes"
else:
    best_mdl = mnb
    best_accuracy = acc_mnb
    model_name = "Multinomial Naive Bayes"

print(f"The best model for the wine dataset is: {model_name} with accuracy: {best_accuracy:.4f}")
```
