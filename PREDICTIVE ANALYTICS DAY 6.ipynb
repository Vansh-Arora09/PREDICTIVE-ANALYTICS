```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.linear_model import LinearRegression
```

```python
# POLYNOMIAL REGRESSION -> train non linear data with curved terms
# In case of linear regression , do all MSE, MAE, R^2, RMSE, although  out of these r^2 is the best
```

```python
df = pd.read_csv('Position_Salaries.csv')
X = df.iloc[:,1:-1].values
print(X)
y = df.iloc[:,-1].values
print(y)
```

```python
# Training the linear regression model
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X,y)
```

```python
# Train the polynomial regression model
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=4)
X_poly = poly_reg.fit_transform(X)
print(X_poly)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X_poly,y)
```

```python
# Vsualizing the Linear Regression Results
```

```python
plt.scatter(X,y,color='red')
plt.plot(X,lin_reg.predict(X),color='blue',linewidth=2)
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

```python
#Visualizing the Polynomial regression Results
```

```python
plt.scatter(X,y,color='red')
plt.plot(X,lin_reg_2.predict(poly_reg.fit_transform(X)),color='blue',linewidth = 2)
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

```python
import seaborn as sns
data = pd.read_csv('titanic.csv')
data
```

```python
sns.countplot(x='Survived',data=data)
sns.countplot(x='Survived',hue='Sex',data=data)
sns.countplot(x ='Survived',hue='Pclass',data=data)
```

```python
data.isnull()[:10]
```

```python
data.isnull().sum()
```

```python
data.drop('Cabin',axis=1,inplace=True)
```

```python
data.dropna(inplace=True)
```

```python
data.head()
```

```python
lb = LabelEncoder()
data['Embarked'] = lb.fit_transform(data['Embarked'])
data['Sex'] = lb.fit_transform(data['Sex'])    # sex = pd.get_dummies(data['sex'],drop_first=True,dtype=int) , drop_first returning sex into 1 column
```

```python
data
```

```python
pcl = pd.get_dummies(data['Pclass'],drop_first=True,dtype=int)
pcl
```

```python
data = pd.concat([data,pcl],axis=1)
data
```

```python
data = data.drop(['Pclass','Name','Ticket','PassengerId'],axis=1)
data
```

```python
X = data.drop('Survived',axis=1)
y = data['Survived']
print(X.head())
print(y.head())
X.rename(columns={2:'Pclass_2',3: 'Pclass_3'},inplace=True)
y.value_counts()
print('Survived' in X.columns)

#before removing the sex column
corr = pd.concat([X,y],axis=1).corr()
print(corr['Survived'].sort_values(ascending=False))

X = X.drop(['Sex'],axis=1)
#After removing the male column
corr = pd.concat([X,y],axis=1).corr()
print(corr['Survived'].sort_values(ascending=False))
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y)
print(X_train.equals(X_test))
print(y_train.equals(y_test))
```

```python
from sklearn.linear_model import LogisticRegression
logmodel = LogisticRegression(max_iter=1000)
logmodel.fit(X_train,y_train)
```

```python
predictor = logmodel.predict(X_test)
predictor
```
